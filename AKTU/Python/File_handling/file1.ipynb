```python
with open ("exFile1.txt" , 'r') as f:
    while True:
        l = f.readline()
        print(l)
        if (l == ''):
            break;
```

```python
with open ("exFile1.txt" , 'w') as f:
    while True:
        l = input("Enter your text: ")
        f.writelines(l)
        f.write('\n')
        if (l == ''):
            break
```

```python
with open ("exFile1.txt" , 'a') as f:
    while True:
        l = input("Enter your text: ")
        f.writelines(l)
        f.write('\n')
        if (l == ''):
            break
```
