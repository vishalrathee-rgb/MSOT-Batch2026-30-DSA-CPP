##### Python caches small integers and strings to optimize memory usage. When you create a new integer or string, Python checks if it already exists in the cache. If it does, Python reuses the existing object instead of creating a new one. This means that two variables with the same value may actually refer to the same object in memory.

###### But, the moment you perform an operation on one of the variables that changes its value, Python creates a new object in memory for that variable, and the two variables will no longer refer to the same object. This is known as "copy-on-write" behavior.

```python
a = 11
b = 11
print(id(a) == id(b))
```

```python
a = 1000
b = 1000
print(id(a) == id(b))
```

```python
a = -126
b = -126
print(id(a) == id(b))
```
