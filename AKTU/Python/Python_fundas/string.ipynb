## <------Strings------>

##### Strings are a sequence of characters enclosed in either single quotes (' ') or double quotes (" "). They are used to represent text in Python. Strings can contain letters, numbers, symbols, and whitespace characters.


##### Strings are immutable, meaning that once a string is created, it cannot be changed. However, you can create new strings by concatenating or modifying existing ones.

###### Multiple quotes can be used to create multi-line strings. Triple quotes (''' or """) allow you to create strings that span multiple lines, preserving the line breaks and formatting.

```python
s1 = "I love Python."
```

```python
s1
```

```python
s2 = """Multi-line text in python.
Python is great.
I am learning strings in python"""
```

```python
s2
```

##### Even though strings are immutable, you can perform various operations on them, such as concatenation, slicing, and formatting. You can also use string methods to manipulate and analyze strings.
