##### Functions are reusable blocks of code that perform a specific task. They allow you to break down your code into smaller, manageable pieces, making it easier to read, maintain, and debug. Functions can take input parameters, perform operations, and return output values.
1. Functions are defined using the `def` keyword, followed by the function name and parentheses. The code block within the function is indented.
2. Functions can take input parameters, which are values passed to the function when it is called. These parameters can be used within the function to perform operations.
3. Functions can return output values using the `return` statement. This allows you to get results from the function and use them in other parts of your code.
4. Functions can be called multiple times from different parts of your code, allowing for code reuse and modular programming.
5. in Python, we don't need to specify the return type of a function, as Python is dynamically typed. The return type is determined by the value returned by the function at runtime.

```python
def nth_fibonacci(n):
    if n <= 0:
        return "Input should be a positive integer."
    elif n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        a, b = 0, 1
        for _ in range(2, n):
            a, b = b, a + b
        return b
```

```python
n1 = int(input("Enter the position of the Fibonacci number: "))
result = nth_fibonacci(n1)
print(f"The {n1}th Fibonacci number is: {result}")
```

```python
def fib_series(n):
    if n <= 0:
        return "Input should be a positive integer."
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        series = [0, 1]
        for _ in range(2, n):
            series.append(series[-1] + series[-2])
        return series
```

```python
print(fib_series(10))
```

#### Default Arguments: In Python, you can define functions with default arguments. Default arguments are values that are assigned to parameters if no value is provided when the function is called. This allows you to call the function with fewer arguments than defined, and the default values will be used for the missing arguments.


```python
def greet(name = "Mirai"):
    return f"Hello, {name}!"
```

```python
greet()  # Uses default argument
```

```python
greet("Ghost")  # Uses default argument
```

###### Function to check if a number is prime or not

```python
def check_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
```

```python
cp = int(input("Enter a number to check if it's prime: "))
if check_prime(cp):
    print(f"{cp} is a prime number.")
else:
    print(f"{cp} is not a prime number.")
```

##### Function that reads the 3 numbers from the user and prints the largest of the three numbers.

```python
def largest_of_three():
    a = float(input("Enter the first number: "))
    b = float(input("Enter the second number: "))
    c = float(input("Enter the third number: "))
    
    largest = max(a, b, c)
    print(f"The largest of the three numbers is: {largest}")
```

```python
largest_of_three()
```

##### print multiplication table of a number using function

```python
def multiplication_table(n):
    print(f"Multiplication table for {n}:")
    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")
```

```python
input_number = int(input("Enter a number to print its multiplication table: "))
multiplication_table(input_number)
```
