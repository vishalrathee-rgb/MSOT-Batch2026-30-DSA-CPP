## <---------Operators in Python--------->

1. Arithmatic Operators: These operators are used to perform mathematical operations on numbers. They include addition (+), subtraction (-), multiplication (*), division (/), modulus (%), exponentiation (**), and floor division (//).

```python
a = 100
b = 189
```

```python
print("Sum of the nos.:",a + b)
print("Product of the nos.:",a * b)
print("Exp of the nos.:",a**b)
print ("Integer division of nos." , a//b)
print("Division: " , a/b)
print("Remainder: ", a%b)
```

2. Assignment Operators: These operators are used to assign values to variables. They include the basic assignment operator (=) as well as compound assignment / shorthand operators such as +=, -=, *=, /=, %=, **=, and //=.

3. Comparison Operators: These operators are used to compare two values and return a boolean result (True or False). They include equal to (==), not equal to (!=), greater than (>), less than (<), greater than or equal to (>=), and less than or equal to (<=).
4. Logical Operators: These operators are used to combine multiple boolean expressions and return a boolean result. They include and, or, and not.

5. Identity & Membership Operators: These operators are used to test the identity of objects and membership in sequences. Identity operators include is and is not, while membership operators include in and not in.

6. Bitwise Operators: These operators are used to perform bit-level operations on integers. They include AND (&), OR (|), XOR (^), NOT (~), left shift (<<), and right shift (>>).    // Extra
