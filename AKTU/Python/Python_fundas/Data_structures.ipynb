# Data Structures in Python

A *data structure* is a way to arrange data so that a program can work with it. The choice affects how we access values, update them, and express a solution. Python's built-in structures—lists, tuples, dictionaries, and sets—cover many everyday tasks. NumPy adds efficient arrays for numerical data.

By the end, you should be able to select a suitable structure, write basic operations and traversals, and connect these ideas to a small sensor-data example.


## 1. Why data structures matter

Suppose we record the temperatures measured during one hour. We could put the readings into one variable each, but that becomes difficult to manage as the number of readings grows. A collection lets us keep related values together and process them with loops.

Different structures have different strengths:

| Structure | Use it when… | Example |
|---|---|---|
| List | order matters and values may change | readings collected over time |
| Tuple | values form a fixed group | (latitude, longitude) |
| Dictionary | values are looked up by a key or label | sensor name → latest reading |
| Set | only distinct values matter | unique device IDs |
| NumPy array | you need efficient numerical calculations | thousands of measurements |

Data structures do not automatically make every program faster or better. They help us represent the problem clearly and use operations that fit the task.


```python
# A collection is easier to process than many separate variables.
readings = [21.4, 21.8, 22.1, 21.9]

for reading in readings:
    print(f"Reading: {reading} °C")
```

## 2. Lists: ordered, changeable collections

A **list** stores items in order. Lists are *mutable*: you can replace, add, and remove items after creating one. They can contain repeated values and, if needed, values of different types; in practice, a list usually represents one kind of thing.

### Indexing and slicing

Python uses **zero-based indexing**: the first item is at index `0`. Negative indexes count backward, so `-1` means the last item. A slice `items[start:stop]` includes `start` and stops just before `stop`.


```python
temperatures = [18.5, 19.0, 19.8, 20.2, 19.7]

print(temperatures[0])      # first item: 18.5
print(temperatures[-1])     # last item: 19.7
print(temperatures[1:4])    # indexes 1, 2, 3 (stop is excluded)
print(temperatures[:3])     # from the beginning up to index 2
print(temperatures[::2])    # every second item
print(temperatures[1:4:2])
```

### Updating lists and useful methods

A **method** is an operation called on an object with dot notation. Common list methods include `append` (add one item at the end), `extend` (add items from another iterable), `insert`, `remove`, `pop`, `sort`, and `count`.

Some methods change the existing list and return `None`. For example, use `values.sort()` to change a list in place; use `sorted(values)` when you want a new sorted list and need to keep the original unchanged.


List methods and built-in functions provide common operations. `len(values)` gives the number of items; `sum`, `min`, and `max` work with numerical lists. `count(value)` returns the number of matches, and `index(value)` returns the first matching position. `clear()` removes all items. `remove(value)` raises `ValueError` if the value is absent; `pop()` raises `IndexError` if there is no item to remove (or the requested index is invalid).


```python
values = [4, 7, 4, 9]
print("Length:", len(values))
print("Sum / minimum / maximum:", sum(values), min(values), max(values))
print("Count of 4:", values.count(4))
print("First index of 9:", values.index(9))
values.clear()
print("After clear:", values)
```

```python
scores = [72, 85, 91]
scores.append(88)
scores.extend([76, 90])
scores.insert(1, 80)
print("After adding:", scores)

removed_score = scores.pop()  # removes and returns the last item
print("Removed:", removed_score)
scores.remove(72)             # removes the first matching value
print("After removing:", scores)
print("Sorted copy:", sorted(scores))
print("Original order:", scores)
```

### Traversing (visiting) a list

**Traversal** means visiting collection items one by one. Use a direct `for item in list` loop when you need values. Use `enumerate` when you also need each position. A running total is called an **accumulator**.


```python
measurements = [3.2, 4.1, 2.9, 5.0]
total = 0.0

for position, value in enumerate(measurements):
    print(f"Position {position}: {value}")
    total += value

average = total / len(measurements)
print("Total:", total)
print("Average:", average)
```

### List comprehension

A **list comprehension** builds a new list from an iterable, often by transforming or filtering each item. The expression below converts each temperature from Celsius to Fahrenheit.


```python
celsius = [0, 10, 20, 30]
fahrenheit = [temperature * 9 / 5 + 32 for temperature in celsius]
print(fahrenheit)

# Keep only readings that meet a condition:
warm_readings = [value for value in celsius if value >= 20]
print(warm_readings)
```

## 3. Tuples: fixed groups of values

A **tuple** is an ordered collection that cannot be changed after it is created. This property is called **immutability**. Tuples are useful when several values belong together and the group should stay fixed, such as an `(x, y)` point or a `(name, value, unit)` record.

Use commas to make a tuple. Parentheses are often used for readability. A one-item tuple needs a trailing comma: `(7,)`; `(7)` is just the integer `7`.


```python
location = (28.6139, 77.2090)  # latitude, longitude
print(location[0])
print(location[-1])
print(type(location))

single_item = (7,)
print(single_item, type(single_item))
```

### Unpacking tuples

**Unpacking** assigns tuple items to several variables in one statement. The number of names must match the number of values. The `_` name conventionally marks a value we do not need.


A tuple may contain a mutable value such as a list. The tuple slot itself cannot be reassigned, but the list inside it can still be changed. This is a useful distinction between an immutable container and the objects it refers to.


```python
record = ("station-1", [20.0, 20.5])
record[1].append(21.0)  # the inner list is mutable
print(record)
```

```python
point = (4, 9)
x, y = point
print("x:", x, "y:", y)

sensor_record = ("sensor-A", 23.6, "°C")
sensor_name, value, unit = sensor_record
print(f"{sensor_name} measured {value} {unit}")

# Swap values with tuple unpacking
x, y = y, x
print("After swap:", x, y)
```

Tuple immutability is shallow: the tuple's slots cannot be reassigned, though a mutable object stored inside a tuple could itself change. For introductory work, the key idea is that a tuple is a good choice for a fixed record, while a list is better for a collection that will be edited.


```python
# A function can return multiple results as a tuple.
def min_and_max(values):
    return min(values), max(values)

lowest, highest = min_and_max([8, 3, 11, 5])
print("Lowest:", lowest, "Highest:", highest)
```

## 4. Dictionaries: key-value pairs

A **dictionary** stores associations between **keys** and **values**. A key is a label used to look up its value. Keys must be unique; assigning a value to an existing key replaces the old value. Dictionaries preserve insertion order in modern Python, but code should use keys rather than rely on a particular display order.

Keys are commonly strings or numbers. Use `dictionary[key]` when the key is known to exist. Use `.get(key, default)` for a safe lookup when it may be missing.


```python
student = {"name": "Asha", "year": 1, "course": "Computer Science"}
print(student["name"])
print(student.get("email", "No email recorded"))

student["year"] = 2            # update a value
student["email"] = "asha@example.edu"  # add a key-value pair
print(student)

for key, value in student.items():
    print(f"{key}: {value}")
```

### Frequency counting with a dictionary

A frequent task is counting how often each item occurs. For every item, increase the count stored under that item as a key. `get(item, 0)` supplies zero the first time that item appears.


Dictionaries are mutable, have unique keys, and preserve insertion order in modern Python. Accessing a missing key with `mapping[key]` raises `KeyError`. `mapping.get(key)` returns `None` by default, or a supplied default value; this makes `.get(key, 0)` useful for frequency counts. Other helpful methods include `keys()`, `values()`, `items()`, `update()`, and `pop()`.


```python
inventory = {"apples": 12, "pears": 5}
print("Known keys:", list(inventory.keys()))
print("Known quantities:", list(inventory.values()))
print("Pairs:", list(inventory.items()))
print("Missing category:", inventory.get("oranges"))
inventory.update({"pears": 7, "plums": 3})
print("Updated:", inventory)
```

```python
letters = "data structures"
frequency = {}

for character in letters:
    if character != " ":  # ignore spaces for this example
        frequency[character] = frequency.get(character, 0) + 1

print(frequency)
print("Number of a's:", frequency.get("a", 0))
```

The same pattern can count categories, words, or sensor states. For example, a dictionary can map each sensor ID to its most recent reading, or map each error code to the number of times it occurred.


```python
status_log = ["ok", "warning", "ok", "offline", "ok", "warning"]
counts = {}
for status in status_log:
    counts[status] = counts.get(status, 0) + 1
print(counts)
```

## 5. Sets: uniqueness and mathematical operations

A **set** is an unordered collection of distinct, hashable items. Adding a duplicate does not create another copy. Use sets to remove duplicates, test membership, or compare groups. An empty set is written `set()`; `{}` creates an empty dictionary.


Sets remove duplicates and support fast membership checks. They are unordered, so they do not support positional indexing such as `items[0]`; their printed order should not be treated as meaningful. Set elements must be hashable. Use `set()` for an empty set because `{}` creates an empty dictionary.


```python
lab_groups = ["A", "B", "A", "C", "B", "A"]
unique_groups = set(lab_groups)
print(unique_groups)
print("A present?", "A" in unique_groups)

unique_groups.add("D")
unique_groups.discard("B")  # discard does not raise an error if absent
print(unique_groups)
```

### Set mathematics

For sets `A` and `B`:

- **Union** (`A | B`): items in either set.
- **Intersection** (`A & B`): items in both sets.
- **Difference** (`A - B`): items in A but not B.
- **Symmetric difference** (`A ^ B`): items in exactly one set.

These operations are useful for comparing membership, such as finding students enrolled in one or both workshops.


```python
python_group = {"Asha", "Ben", "Chen", "Diya"}
robotics_group = {"Ben", "Diya", "Eli"}

print("Either workshop:", python_group | robotics_group)
print("Both workshops:", python_group & robotics_group)
print("Python only:", python_group - robotics_group)
print("Exactly one:", python_group ^ robotics_group)
```

## 6. NumPy arrays: numerical data

[NumPy](https://numpy.org/) is a Python library for working with numerical arrays. An array can hold many numbers in a compact structure and apply arithmetic to all values at once. Unlike a Python list, NumPy arithmetic is generally **element-wise**: adding a number to an array adds it to every element.

The next cells import NumPy as `np`. If this import fails in your Python environment, install NumPy there before running these examples.


```python
import numpy as np

samples = np.array([20.1, 20.4, 20.0, 20.8])
print(samples)
print("Type:", type(samples))
print("Shape:", samples.shape)
print("Number of dimensions:", samples.ndim)
print("Data type:", samples.dtype)
```

### Shape and dimensions

An array's **shape** tells the length along each dimension. A one-dimensional array of four values has shape `(4,)`. A two-dimensional array might represent rows of sensors and columns of time steps. Keep track of dimensions so you know what each axis represents.


For NumPy arrays, `ndim` is the number of dimensions, `shape` gives the length of each axis, and `size` is the total number of elements. For example, shape `(2, 3)` means two rows and three columns, for six values in total. In two-dimensional indexing, write the row index first and the column index second.


```python
table = np.array([[10, 20, 30], [40, 50, 60]])
print("Dimensions:", table.ndim)
print("Shape:", table.shape)
print("Total elements:", table.size)
print("Row 0, column 1:", table[0, 1])
print("First row:", table[0])
```

```python
# Rows can represent sensors; columns can represent time steps.
readings = np.array([
    [20.1, 20.4, 20.0, 20.8],  # sensor A
    [18.9, 19.2, 19.4, 19.1],  # sensor B
])
print(readings)
print("Shape (rows, columns):", readings.shape)
print("Sensor A, third time step:", readings[0, 2])
print("First sensor row:", readings[0, :])
print("Second time step for all sensors:", readings[:, 1])
```

### Arithmetic and mean

NumPy operations can transform a whole array without an explicit Python loop. Use `np.mean` to calculate an average. For a two-dimensional array, `axis=0` computes a result for each column and `axis=1` computes a result for each row.


```python
readings = np.array([
    [20.1, 20.4, 20.0, 20.8],
    [18.9, 19.2, 19.4, 19.1],
])

calibrated = readings + 0.5
fahrenheit = readings * 9 / 5 + 32
print("Calibrated values:\n", calibrated)
print("Fahrenheit:\n", fahrenheit)
print("Mean of every measurement:", np.mean(readings))
print("Mean per sensor (row):", np.mean(readings, axis=1))
print("Mean per time step (column):", np.mean(readings, axis=0))
```

### Lists and arrays compared

A list is a general-purpose Python collection. A NumPy array is designed for numerical work and supports vectorized arithmetic and multidimensional shapes. For example, `list_a + list_b` joins two lists, while adding NumPy arrays of matching shape adds corresponding elements. Choose the structure based on the operations your task needs.


The arithmetic difference is important: multiplying a Python list by `2` repeats its contents, while multiplying a NumPy array by `2` doubles every element. A list cannot add a scalar with `+`; a NumPy array can. Operations between arrays are element-wise when their shapes are compatible.


```python
values_list = [2, 4]
values_array = np.array(values_list)
print("List repeated:", values_list * 2)
print("Array doubled:", values_array * 2)
print("Array plus scalar:", values_array + 3)
print("Arrays added element-wise:", values_array + np.array([10, 20]))
```

```python
list_a = [1, 2, 3]
list_b = [10, 20, 30]
print("List + list:", list_a + list_b)  # concatenation

array_a = np.array(list_a)
array_b = np.array(list_b)
print("Array + array:", array_a + array_b)  # element-wise addition
```

## 7. Bridge to sensor-data analysis

A simple sensor pipeline can combine several structures:

1. Store a time-ordered stream of readings in a **list**.
2. Store a fixed reading and its unit as a **tuple**.
3. Store device information or summary values in a **dictionary**.
4. Use a **set** to find distinct sensor IDs.
5. Convert numeric readings to a **NumPy array** for calculations.

The example below uses made-up temperature readings. Real sensor systems also need to consider missing readings, timestamps, calibration, noise, and measurement units.


A basic sensor summary often includes the mean, highest value, and lowest value. These describe the central level and observed range of the readings. A later step might plot readings against time; plotting is outside this lesson, but the organized numerical data makes it possible.


```python
# Each record is: (sensor ID, time in seconds, temperature in Celsius)
records = [
    ("S1", 0, 21.1),
    ("S2", 0, 19.8),
    ("S1", 1, 21.4),
    ("S2", 1, 20.0),
    ("S1", 2, 21.2),
    ("S2", 2, 20.3),
]

sensor_ids = {record[0] for record in records}
print("Sensors:", sensor_ids)

# Group temperatures by sensor using a dictionary of lists.
by_sensor = {}
for sensor_id, time_seconds, temperature_c in records:
    by_sensor.setdefault(sensor_id, []).append(temperature_c)

print("Grouped readings:", by_sensor)

# Summarize each sensor's readings with NumPy.
summary = {}
for sensor_id, values in by_sensor.items():
    values_array = np.array(values)
    summary[sensor_id] = {
        "count": len(values_array),
        "mean_c": np.mean(values_array),
        "minimum_c": np.min(values_array),
        "maximum_c": np.max(values_array),
    }

print("Summary:")
for sensor_id, stats in summary.items():
    print(sensor_id, stats)
```

## 8. Choosing a suitable structure

Use a list for a sequence that may change, a tuple for a fixed record, a dictionary for key-value lookup, a set for unique values, and a NumPy array for numerical computation. A single program can combine these structures when that makes the data easier to represent and analyze.
