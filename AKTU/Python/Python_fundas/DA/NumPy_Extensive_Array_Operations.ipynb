# NumPy Array Operations

This notebook is a practical guide to common NumPy operations. It builds on the first array examples and covers creating arrays, inspecting their shape, selecting values, changing their layout, performing calculations, filtering data, and summarizing measurements.

NumPy is designed for numerical data. Its operations can work on a whole array at once, which makes code shorter and often more efficient than processing each number in a Python loop. Run the cells in order.


## 1. Import NumPy

NumPy is usually imported with the short name `np`. Install it in the Python environment used by Jupyter if needed: `pip install numpy`.


```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 2. Create arrays

`np.array` converts a list or nested lists into an array. NumPy also provides functions for regular sequences, evenly spaced values, and arrays filled with a constant. Choose the constructor that makes the intended values clear.


```python
from_list = np.array([2, 4, 6, 8])
sequence = np.arange(0, 10, 2)        # start, stop (excluded), step
even_points = np.linspace(0, 1, 5)   # five evenly spaced values, including endpoints
zeros = np.zeros(3)
ones = np.ones((2, 3))
constant = np.full((2, 2), 7)
identity = np.eye(3)

print("From list:", from_list)
print("arange:", sequence)
print("linspace:", even_points)
print("Zeros:", zeros)
print("Ones:\n", ones)
print("Constant values:\n", constant)
print("Identity matrix:\n", identity)
```

`arange` is useful for integer steps. Floating-point steps can accumulate rounding differences, so `linspace` is often a better choice when you want a fixed number of points across an interval.


## 3. Inspect array properties

An array's `ndim` is the number of dimensions, `shape` gives the length along each axis, `size` is the total number of elements, and `dtype` is the element data type. NumPy arrays usually store values using one common type.


```python
matrix = np.array([[10, 20, 30], [40, 50, 60]])
print("Array:\n", matrix)
print("Dimensions:", matrix.ndim)
print("Shape:", matrix.shape)
print("Total values:", matrix.size)
print("Element type:", matrix.dtype)
```

The matrix has shape `(2, 3)`: two rows and three columns. Its size is six because `2 * 3 = 6`. A one-dimensional array with four values has shape `(4,)`; it is not the same shape as a two-dimensional row with shape `(1, 4)`.


## 4. Index and slice arrays

Indexing starts at zero. In a two-dimensional array, provide the row index first and the column index second. Slices use the familiar `start:stop:step` pattern, with the stop position excluded.


```python
values = np.array([5, 10, 15, 20, 25, 30])
print("First:", values[0])
print("Last:", values[-1])
print("Middle slice:", values[1:4])
print("Every second value:", values[::2])

table = np.array([[10, 20, 30], [40, 50, 60]])
print("Row 1, column 2:", table[1, 2])
print("First row:", table[0, :])
print("Second column:", table[:, 1])
print("Top-left 2 by 2 block:\n", table[:2, :2])
```

A slice usually gives a **view** of the original array, so changing an item through the slice can also change the original data. Use `.copy()` when you need an independent array.


```python
original = np.array([1, 2, 3, 4])
view = original[1:3]
view[0] = 99
print("Original after changing a view:", original)

original = np.array([1, 2, 3, 4])
independent = original[1:3].copy()
independent[0] = 99
print("Original after changing a copy:", original)
print("Independent copy:", independent)
```

## 5. Change values and array layout

Arrays are mutable, so individual elements can be updated. `reshape` changes the shape while preserving the element order; the new shape must have the same total number of elements. `ravel` and `flatten` turn a multidimensional array into one dimension. `flatten` makes a copy, while `ravel` may return a view. `.T` transposes rows and columns.


```python
values = np.arange(1, 7)
values[0] = 100
reshaped = values.reshape(2, 3)
print("Updated one-dimensional array:", values)
print("Reshaped:\n", reshaped)
print("Flattened with ravel:", reshaped.ravel())
print("Flattened with flatten:", reshaped.flatten())
print("Transpose:\n", reshaped.T)
```

## 6. Element-wise arithmetic

Arithmetic between arrays of the same shape is performed element by element. A scalar, such as `3`, is applied to every element. `*` means element-wise multiplication; it does not multiply matrices.


```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Element-wise multiplication:", a * b)
print("Division:", a / b)
print("Power:", b ** 2)
print("Add scalar:", a + 5)
print("Multiply by scalar:", a * 2)
```

NumPy also provides element-wise mathematical functions. These are often called **universal functions** (ufuncs). They accept arrays and calculate each result without requiring a loop written by the programmer.


```python
angles = np.array([0, np.pi / 2, np.pi])
print("Square root:", np.sqrt(np.array([1, 4, 9])))
print("Absolute value:", np.abs(np.array([-3, 0, 3])))
print("Sine:", np.sin(angles))
print("Rounded values:", np.round(np.array([2.345, 5.678]), 2))
```

## 7. Compare values and filter with masks

Comparisons such as `>`, `==`, and `<=` return a Boolean array of `True` and `False` values. Use that array as a **mask** to select only values that meet a condition. Combine multiple conditions with `&` for AND and `|` for OR; put each condition in parentheses.


```python
temperatures = np.array([18.2, 21.5, 25.0, 28.4, 30.1])
mask = temperatures >= 25
print("Mask:", mask)
print("At least 25 °C:", temperatures[mask])
print("Between 20 and 28 °C:", temperatures[(temperatures >= 20) & (temperatures <= 28)])
print("Any hot reading:", np.any(temperatures > 30))
print("All readings above 0:", np.all(temperatures > 0))
```

`np.where(condition, value_if_true, value_if_false)` selects or labels values based on a condition. It can create a new result without changing the original array.


```python
temperatures = np.array([18.2, 21.5, 25.0, 28.4, 30.1])
labels = np.where(temperatures >= 25, "warm", "mild")
print(labels)
```

## 8. Broadcasting

**Broadcasting** lets NumPy combine arrays with different but compatible shapes. A scalar can be broadcast across an array. Dimensions are compatible when they are equal or one of them is `1`. This makes it possible to add a calibration offset to each sensor column in a table.


```python
# Rows are time steps; columns are sensors.
readings = np.array([[20.0, 21.0, 19.5], [20.5, 21.5, 20.0]])
calibration_offsets = np.array([0.2, -0.1, 0.5])
calibrated = readings + calibration_offsets
print("Original shape:", readings.shape)
print("Offset shape:", calibration_offsets.shape)
print("Calibrated readings:\n", calibrated)
```

The three offsets line up with the three sensor columns and are applied to each row. Incompatible shapes raise a `ValueError`; checking `.shape` is a useful first step when an array operation fails.


## 9. Aggregations and axes

An **aggregation** reduces many values to a summary, such as a sum or mean. Without an `axis` argument, NumPy reduces the entire array. With `axis=0`, it works down rows and produces one value per column. With `axis=1`, it works across columns and produces one value per row.


```python
measurements = np.array([[20.0, 21.0, 19.0], [22.0, 23.0, 21.0]])
print("Sum of all values:", np.sum(measurements))
print("Mean of all values:", np.mean(measurements))
print("Mean per column:", np.mean(measurements, axis=0))
print("Mean per row:", np.mean(measurements, axis=1))
print("Minimum per column:", np.min(measurements, axis=0))
print("Maximum per row:", np.max(measurements, axis=1))
print("Median:", np.median(measurements))
print("Standard deviation:", np.std(measurements))
print("Index of smallest value in flattened array:", np.argmin(measurements))
```

Useful aggregations include `sum`, `mean`, `median`, `min`, `max`, `std`, `var`, `argmin`, and `argmax`. `cumsum` returns running totals; `diff` returns differences between neighboring values. These are useful for ordered measurements such as sensor readings over time.


```python
daily_steps = np.array([1200, 1800, 1500, 2200])
print("Running total:", np.cumsum(daily_steps))
print("Change from previous day:", np.diff(daily_steps))
print("Index of largest day:", np.argmax(daily_steps))
```

## 10. Sorting and finding distinct values

`np.sort` returns a sorted copy. `np.argsort` returns the positions that would sort the values. `np.unique` returns distinct values in sorted order; it can also return how often each value appears.


```python
scores = np.array([82, 95, 82, 71, 95, 88])
print("Sorted copy:", np.sort(scores))
print("Original:", scores)
print("Order of sorted positions:", np.argsort(scores))
unique_scores, counts = np.unique(scores, return_counts=True)
print("Distinct scores:", unique_scores)
print("Counts:", counts)
```

## 11. Join and split arrays

`concatenate` joins arrays along an existing axis. `vstack` adds rows and `hstack` joins columns for compatible shapes. `split` divides an array into sections. The arrays being joined must have matching dimensions except along the axis where values are added.


```python
morning = np.array([18, 19, 20])
evening = np.array([22, 21, 20])
print("Concatenated:", np.concatenate((morning, evening)))

first_day = np.array([[18, 19], [20, 21]])
second_day = np.array([[22, 23], [24, 25]])
print("Stacked rows:\n", np.vstack((first_day, second_day)))
print("Stacked columns:\n", np.hstack((first_day, second_day)))
left, right = np.split(np.array([1, 2, 3, 4, 5, 6]), 2)
print("Split parts:", left, right)
```

## 12. Matrix multiplication

For two-dimensional arrays, `@` performs matrix multiplication. This differs from `*`, which multiplies corresponding elements. For matrix multiplication, the number of columns in the left array must equal the number of rows in the right array.


```python
left_matrix = np.array([[1, 2], [3, 4]])
right_matrix = np.array([[10, 20], [30, 40]])
print("Element-wise product:\n", left_matrix * right_matrix)
print("Matrix product:\n", left_matrix @ right_matrix)
```

## 13. Missing numerical values

NumPy represents a missing floating-point value with `np.nan`. Ordinary arithmetic with `nan` often produces `nan`, and comparisons with it do not behave like ordinary numeric comparisons. Use `np.isnan` to identify missing values. Functions such as `np.nanmean` calculate a mean while ignoring `nan` values.


```python
sensor_values = np.array([21.0, np.nan, 22.5, 20.5])
missing_mask = np.isnan(sensor_values)
print("Missing-value mask:", missing_mask)
print("Number of missing readings:", np.sum(missing_mask))
print("Mean ignoring missing readings:", np.nanmean(sensor_values))
```

Ignoring missing values is a choice that should be stated. It is not always appropriate to discard them; the reason a measurement is absent may matter to the analysis.


## 14. Reproducible random data

Random numbers are useful for simulations and practice data. `default_rng` creates a random number generator. Giving it a seed makes the same sequence appear each time, which helps students compare results and reproduce examples.


```python
rng = np.random.default_rng(seed=42)
random_readings = rng.normal(loc=22.0, scale=1.5, size=5)
print("Simulated readings:", np.round(random_readings, 2))
```

## 15. Sensor-data example

This final example stores readings as rows and sensor locations as columns. The array operations convert Celsius to Fahrenheit, calculate average values, and identify temperatures above a chosen threshold. The threshold is simply part of this demonstration; a real monitoring system would set it based on the device and use case.


```python
# Each row is a time step; each column is a sensor location.
readings_c = np.array([
    [20.1, 21.3, 19.8],
    [20.8, 22.0, 20.2],
    [21.4, 22.5, 20.7],
    [21.0, 22.1, 20.4],
])

readings_f = readings_c * 9 / 5 + 32
mean_per_sensor = np.mean(readings_c, axis=0)
maximum_per_time = np.max(readings_c, axis=1)
hot_readings = readings_c[readings_c >= 22.0]

print("Shape (time steps, sensors):", readings_c.shape)
print("Fahrenheit readings:\n", np.round(readings_f, 1))
print("Mean per sensor (°C):", np.round(mean_per_sensor, 2))
print("Maximum at each time step (°C):", maximum_per_time)
print("Readings at or above 22 °C:", hot_readings)
```

## Summary

NumPy arrays can be created from data or with helper functions. Their shape and axes describe how values are arranged. Indexing, slicing, masks, and reshaping select or reorganize values. Arithmetic, mathematical functions, broadcasting, and matrix multiplication perform calculations. Aggregations reduce data to summaries, while sorting, joining, and splitting help organize it. Always check dimensions, data types, missing values, and the meaning of each axis.
