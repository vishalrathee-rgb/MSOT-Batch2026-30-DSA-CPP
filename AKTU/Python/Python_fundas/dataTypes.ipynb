## Data Types in Python

#### Python has several built-in data types that are used to store different kinds of data. The most commonly used data types in Python are:


1. Numeric Types: These include integers (int), floating-point numbers (float), and complex numbers (complex). Examples:
   - int: 5, -10, 0
   - float: 3.14, -2.5, 0.0
   - complex: 2 + 3j, -1 - 4j
2. Sequence Types: These include strings (str), lists (list), and tuples (tuple). Examples:
   - str: "Hello", 'Python', "123"
   - list: [1, 2, 3], ['a', 'b', 'c'], [1, 'a', 3.14]
   - tuple: (1, 2, 3), ('x', 'y', 'z'), (1, 'a', 3.14)
3. Mapping Type: This includes dictionaries (dict), which store key-value pairs. Example:
   - dict: {'name': 'Alice', 'age': 25}, {1: 'one', 2: 'two'}
4. Set Types: These include sets (set) and frozensets (frozenset), which are unordered collections of unique elements. Examples:
   - set: {1, 2, 3}, {'a', 'b', 'c'}
   - frozenset: frozenset([1, 2, 3]), frozenset({'x', 'y', 'z'})
5. Boolean Type: This includes the bool type, which can have two values: True or False. Example:
   - bool: True, False 


#### INT --> int itself has several subtypes based on the size of the integer. The most commonly used subtypes are:
- int: This is the standard integer type in Python. It can represent both positive and negative whole numbers. Example: 5, -10, 0
- long: In Python 2, there was a separate long type for representing larger integers.
#### Python allows you to perform operations on 4 systems of numbers:
1. Binary (base 2): Uses only the digits 0 and 1. Example: 0b1010 (binary representation of 10)
2. Octal (base 8): Uses digits from 0 to 7. Example: 0o12 (octal representation of 10)
3. Decimal (base 10): The standard number system we use in everyday life. Example
4. Hexadecimal (base 16): Uses digits from 0 to 9 and letters A to F. Example: 0xA (hexadecimal representation of 10)

##### basic operations on decimal:

```python
i1 = 10
i2 = 3
print("sum of " , i1 , " and " , i2 , " is: " , i1 + i2 , ". ")
```

##### basic operation on binary:

```python
b1 = 0b0011
b2 = 0b1101
```

```python
print("sum of " , b1 , " and " , b2 , " is: " , b1 + b2, ". ")
```

##### basic operation on octal:

```python
o1 = 0o74266
o2 = 0o35427
print("The sum of " , o1 , "and" , o2, "is: " , o1 + o2)
```

##### basic operation on hexadecimal:

```python
h1 = 0xA272B9
h2 = 0xB9477C
print("The sum of " , h1 , "and" , h2, "is: " , h1 + h2)
```
